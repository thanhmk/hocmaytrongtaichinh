## **1. Khung Bayes**

Ta thường muốn tìm xác suất một quan sát nhận một giá trị nhóm nhất định, chẳng hạn thành công/thất bại hay mua/giữ, là ví dụ của bài toán phân loại nhị phân. Những sự kiện này gọi là một tập sự kiện đầy đủ và loại trừ nhau, trong đó chỉ có hai kết quả có thể xảy ra, chẳng hạn $A$ và $\bar{A}$, $P(A)+P(\bar{A})=1$. Do đó, nếu biết xác suất của một kết quả, ta có thể suy ra xác suất của kết quả kia bằng cách lấy phần bù, tức là $P(A)=1-P(\bar{A})$.

Xác suất của một sự kiện có thể phụ thuộc vào việc một sự kiện khác có xảy ra hay không. Chẳng hạn, ta có thể xét xác suất một sinh viên đậu kỳ thi khi biết sinh viên đó đã học hơn 40 giờ. Trong trường hợp này, xác suất đậu kỳ thi phải cao hơn so với khi sinh viên chỉ học bốn giờ. Nếu ký hiệu các sự kiện:

* $S$ - Sự kiện sinh viên đậu kỳ thi
* $A$ - Sự kiện sinh viên có chuẩn bị cho kỳ thi

Ta có thể viết ký hiệu thống kê $P(S|A)$ là xác suất sinh viên đậu kỳ thi khi biết họ đã chuẩn bị. Đây gọi là **xác suất có điều kiện** và là nền tảng của thống kê Bayes.

**Xác suất đồng thời** là xác suất các sự kiện cùng xảy ra, tức $P(A\cap B)$ là xác suất cả sự kiện A và B cùng xảy ra. Từ ví dụ trên, $P(A\cap S)$ sẽ là xác suất sinh viên chuẩn bị cho kỳ thi và đậu, tức là xác suất đồng thời của A và S.

### **1.1 Xác suất có điều kiện**

Với hai sự kiện bất kỳ $A$ và $B$, quy tắc xác suất có điều kiện là
$$P(A|B) = \frac{P(A\cap B)}{P(B)}\tag{1.1}$$

Ta dùng quy tắc trên trong một ví dụ. Một mẫu được lấy từ nhân viên ngành khai khoáng và ngành tài chính, giữ chức vụ quản lý hoặc giám đốc. Bảng 1 thể hiện mẫu dưới dạng bảng chéo (contingency table). Dựa trên mẫu, giả sử ta muốn biết xác suất một nhân viên là quản lý khi biết họ thuộc ngành tài chính. Dùng (1.1), ta có

$$
\begin{align*}
P(Management|Finance) &= \frac{P(Management\cap Finance)}{P(Finance)}\\
&= \frac{700/1600}{900/1600}\\
&= 0.78
\end{align*}
$$

**Bảng 1**

| | Khai khoáng | Tài chính | TỔNG |
|:---|---:|---:|---:|
| Quản lý | 580 | 700 | 1.280 |
| Giám đốc | 120 | 200 | 320 |
| TỔNG | 700 | 900 | 1.600 |

Giờ ta đã hiểu lý thuyết xác suất cơ bản cần cho cách tiếp cận thống kê Bayes. Ta sẽ bắt đầu bằng định lý Bayes và bộ phân loại Naive Bayes dựa trên định lý này.

### **1.2 Định lý Bayes**

Sắp xếp lại (1.1), ta có xác suất đồng thời là $P(A\cap B)=P(A|B)P(B)$.

Lưu ý rằng sự kiện $A$ và $B$ cũng giống sự kiện $B$ và $A$; do đó biểu thức trên cũng có thể viết là $P(A\cap B)=P(B|A).P(A)$ (1.2).

Thay (1.2) vào (1.1), ta được
$$P(A|B)=\frac{P(B|A)P(A)}{P(B)}\tag{1.3}$$

Phương trình (1.3) gọi là định lý Bayes, là nền tảng của hầu hết các thuật toán dùng cách tiếp cận Bayes. Phương trình (1.3) có thể mở rộng cho nhiều hơn một biến điều kiện. Nói cách khác, thay vì chỉ $B$, ta có thể có $X_1$, $X_2$, ..., $X_n$. Ta viết $X$ thay vào để nhất quán với ký hiệu của các biến giải thích, và dùng $Y$ thay cho $A$ vì biến mục tiêu hay biến phản hồi thường được viết như vậy. Ta cần ký hiệu này để giải thích phương pháp của bộ phân loại Naive Bayes. Giờ ta đã thấy định lý Bayes, ta hãy dùng nó để xây dựng một bộ phân loại.

### **1.3 Naive Bayes**

Giả sử ta có một biến mục tiêu nhận các giá trị nhóm. Khi $Y$ nhận một giá trị nhóm cụ thể, chẳng hạn $k$, phương trình (1.3) có thể viết thành (1.4):

$$P(Y=k|X_1,X_2,\cdots,X_n)=\frac{P(X_1|Y=k)P(X_2|Y=k)\cdots P(X_n|Y=k)P(Y=k)}{P(X_1)P(X_2)\cdots P(X_n)}\tag{1.4}$$

Cần lưu ý rằng phương trình (1.4) đúng với điều kiện các biến giải thích $X_1$, $X_2$, ..., $X_n$ độc lập có điều kiện với $Y$, tức là $P(X_1,X_2,\cdots,X_n|Y=k)=\prod_{i=1}^{n}P(X_i|Y=k)$. Đây là giả định "naive" (ngây thơ) của bộ phân loại, và là lý do cho cái tên. Naive Bayes giả định độc lập có điều kiện, trong khi định lý Bayes thì không. Mẫu số cũng có thể viết là

$$P(X_1\cdots X_n)=\sum_{j}P(Y=j)\prod_{i=1}^{n}P(X_i|Y=j)$$

bằng cách lấy tổng trên mọi nhóm có thể có của biến mục tiêu. Đây là một tính chất của phép lấy biên (marginalization). Các số hạng trong (1.4) có tên gọi riêng, cụ thể là:

* $P(X_1|Y=k).P(X_2|Y=k)\ldots P(X_n|Y=k)=likelihood$ (hợp lý)
* $P(Y=k)=Prior$ (tiên nghiệm)
* $P(X_1,X_2\ldots X_n)=Evidence$ (bằng chứng)
* $P(Y=k|X_1,X_2,\ldots,X_n)=Posterior$ (hậu nghiệm)

nên ta có (1.5):

$$Posterior=\frac{Likelihood\times Prior}{Evidence}\tag{1.5}$$

### **1.4 Ví dụ về Naive Bayes**

Ta xét một ví dụ để hiểu bộ phân loại. Bảng 2 chứa dữ liệu rủi ro tín dụng, cụ thể là những người vỡ nợ và không vỡ nợ thẻ tín dụng trên các khoản vay. Có hai biến giải thích là $X_1$ và $X_2$, lần lượt là ngành nghề làm việc và việc thời gian làm việc có hơn 3 năm hay không. Xác định xác suất vỡ nợ khoản vay là một trọng tâm quan trọng trong ngành ngân hàng.

**Bảng 2**

| | $X_1$ Khai khoáng | $X_1$ Tài chính | $X_1$ Nông nghiệp | $X_2$ Làm việc > 3 năm | $X_2$ Làm việc ≤ 3 năm | TỔNG |
|:---|---:|---:|---:|---:|---:|---:|
| Vỡ nợ (y=1) | 250 | 600 | 150 | 300 | 700 | 1.000 |
| Không vỡ nợ (y=0) | 1.250 | 6.000 | 1.750 | 1.500 | 7.500 | 9.000 |
| TỔNG | 1.500 | 6.600 | 1.900 | 1.800 | 8.200 | 10.000 |

Nếu muốn biết hàm hợp lý cho một nhân viên ngành khai khoáng khi biết họ đã vỡ nợ, ta dùng (1.1) để được (1.6):

$$P(X_1=Mining|Y=1)=\frac{250}{1000}=0.25\tag{1.6}$$

và tương tự, hàm hợp lý của việc làm việc từ 3 năm trở xuống khi biết nhân viên đã vỡ nợ là

$$P(X_2=Employed3yrsunder|Y=1)=\frac{700}{1000}=0.7$$

Bây giờ ta có thể xác định xác suất Naive Bayes của việc vỡ nợ khi biết nhân viên thuộc ngành khai khoáng và đã làm việc từ ba năm trở xuống, dùng (1.4). Để đơn giản, đặt M = Khai khoáng (Mining) và Eu = Làm việc từ 3 năm trở xuống (Employed 3 years or less).

$$
\begin{align*}
P(Y=1|X_1=M,X_2=Eu)&=\frac{P(X_1=M|Y=1)\cdot P(X_2=Eu|Y=1)\cdot P(Y=1)}{P(X_1=M)\cdot P(X_2=Eu)}\\
&=\frac{(0.25)(0.7)(1000/10000)}{(9000/10000)(1250/9000)(7500/9000)+(1000/10000)(250/1000)(700/1000)}=0.14.
\end{align*}
$$

Tương tự, $P(Y=0|X_1=Mining,X_2=Employed3yrsorless)=0.85$. Bộ phân loại sẽ xếp khách hàng này vào nhóm không vỡ nợ vì họ có xác suất không vỡ nợ cao hơn. Có thể tìm tài liệu tham khảo hữu ích để biết thêm chi tiết về phân tích Bayes ở chương 12 của tài liệu của Downey.

Ta đã chỉ ra cách bộ phân loại Naive Bayes hoạt động và các giả định được đưa ra. Bộ phân loại cũng có thể mở rộng cho các biến dự báo liên tục với giả định phân phối chuẩn.
