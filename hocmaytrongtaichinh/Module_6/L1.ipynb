## **1. Học sâu**

Học sâu là một lĩnh vực của học máy được lấy cảm hứng từ hoạt động của mạng nơ-ron trong não người (Zhang et al). Những năm gần đây, học sâu thu hút rất nhiều trường hợp sử dụng nhờ khả năng khái quát hóa mạnh và sức huấn luyện bền vững. Khi dữ liệu rất lớn, học sâu đã được chứng minh là hoạt động tốt hơn nhiều so với các thuật toán học máy cổ điển.

Học sâu đã được áp dụng thành công ở cả bài toán phân loại và hồi quy trong thị giác máy tính, xử lý ngôn ngữ tự nhiên, nhận dạng nghe nhìn và xử lý ảnh. Gần đây, các ứng dụng học sâu đã được nghiên cứu và triển khai trong các lĩnh vực tài chính, y học, vận trù học và ngân hàng. Việc ứng dụng học sâu trong tài chính được thúc đẩy bởi sự tồn tại của khối lượng dữ liệu tài chính khổng lồ, vốn không đầy đủ, không chắc chắn và không nhất quán, gây ra thách thức lớn cho các phương pháp thống kê truyền thống từng dùng để phân tích dữ liệu tài chính.

Các ứng dụng của học sâu trong thế giới tài chính gồm:
- Mô hình hóa rủi ro tín dụng (Addo et al. 38; Modarres et al.)

Sự tiến bộ của học sâu và học máy trong lĩnh vực tài chính phần lớn được tạo điều kiện bởi:

- Khả năng tiếp cận công nghệ.
- Các đột phá đạt được ở các lĩnh vực khác của học sâu như trò chơi, xử lý ngôn ngữ tự nhiên, xử lý ảnh, v.v.

Trong kỹ thuật tài chính, các bài toán sau có thể giải bằng thuật toán học sâu:
- Phòng ngừa rủi ro sâu (Buehler et al. 1271-1291)

### **1.1 Nơ-ron**

Nơ-ron là đơn vị tính toán cơ bản của mạng nơ-ron nhân tạo. Dưới đây ta xem cách hoạt động của mạng nơ-ron nhân tạo, giả sử các biến dự báo là $x_1$, $x_2$, $x_3$ và $x_4$ và ta đang dự báo $y$.

Vì các biến thường không quan trọng như nhau trong việc dự báo đầu ra, ta gán các hệ số nhân trọng số cho mọi đầu vào để thể hiện mức quan trọng. Do đó, cộng các đầu vào đã nhân với trọng số sẽ cho
$$x_1 \times w_1 + x_2 \times w_2 + x_3 \times w_3 + x_4 \times w_4$$
Thêm số hạng chệch vào phương trình trên, ta được
$$z = (x_1 \times w_1 + x_2 \times w_2 + x_3 \times w_3 + x_4 \times w_4) + b$$

Phương trình trên giống phương trình hồi quy tuyến tính. Điểm khác giữa nơ-ron và hồi quy tuyến tính là việc đưa vào một hàm phi tuyến gọi là hàm kích hoạt, áp dụng lên $z$ để dự báo đầu ra, tức là, $$y = f(z)$$

## **1.3 Hàm kích hoạt**

Như đã chỉ ra ở trên, hàm kích hoạt được dùng để đưa tính phi tuyến vào đầu vào nhằm học mối quan hệ nền tảng trong dữ liệu, qua đó cho phép khớp phức tạp. Nếu không có phép biến đổi phi tuyến, mạng nơ-ron chỉ là một mô hình hồi quy tuyến tính.

### **1.3.1 Hàm Sigmoid (hàm logistic)**
Hàm sigmoid được định nghĩa là $$f(x) = \frac{1}{1+e^{-x}}$$
Nó cho một đường cong hình chữ $S$ như bên dưới:

```python
import math

import matplotlib.pyplot as plt
import numpy as np


def sigmoid(x):
    a = []
    for item in x:
        a.append(1 / (1 + math.exp(-item)))
    return a


x = np.arange(-5.0, 5.0, 0.2)
sig = sigmoid(x)

print("\033[1m" + "Fig 3: A sigmoid function curve" + "\033[0m")
plt.plot(x, sig)
# `plt.suptitle`("Fig. 3: A sigmoid function curve")
plt.show()
```

Hàm sigmoid khả vi và đơn điệu, và cho giá trị đầu ra giữa $0$ và $1$. Ta có thể viết hàm sigmoid trong Python như sau:

```python
import numpy as np


def sigmoid(x):
    return 1 / (1 + np.exp(-x))
```

#### **1.3.2 Hàm Tanh**
Hàm tanh biểu diễn là $$f(x) = \frac{1 - e^{-2x}}{1+e^{-2x}}$$ giống hàm sigmoid ở trên, điểm khác là nó cho giá trị đầu ra giữa $-1$ và $1$ như bên dưới.

```python
import matplotlib.pyplot as plt


def tanh(x):
    a = []
    for item in x:
        a.append((1 - math.exp(-2 * item)) / (1 + math.exp(-2 * item)))
    return a


x = np.arange(-5.0, 5.0, 0.2)
tanva = tanh(x)

print("\033[1m" + "Fig 4: A tanh function curve" + "\033[0m")
plt.plot(x, tanva)
plt.show()
```

Có thể cài đặt trong Python như bên dưới:

```python
def tanh(x):
    return (1 - np.exp(-2 * x)) / (1 + np.exp(-2 * x))
```

#### **1.3.3 Hàm Rectified Linear Unit (ReLU)**
Hàm ReLU biểu diễn toán học là
$$
f(x) = \begin{cases} 
      0 & \text{với }x< 0 \\
      x & \text{với }x\geq 0
 \end{cases}
 $$
 hay $$f(x) = \max\{0, x\}$$
 có đầu ra là các số không âm.
 
 Có thể hình dung bằng hình ảnh như bên dưới.

```python
import numpy as np

x = np.linspace(-10, 10, 1000)
y = np.maximum(0, x)

print("\033[1m" + "Fig 5: A ReLU function curve" + "\033[0m")
plt.figure(figsize=(10, 5))
plt.plot(x, y)
plt.legend(["Relu"])
# `plt.suptitle`("Fig. 5: A ReLU function curve")
plt.show()
```

Mã Python cho ReLU là:

```python
def ReLU(x):
    if x < 0:
        return 0
    else:
        return x
```

#### **1.3.4 Hàm Leaky ReLU**
Leaky ReLU là bản điều chỉnh của hàm ReLU nhằm giải quyết vấn đề ReLU chết (dying ReLU). Biểu diễn toán học là:
$$
f(x) = \begin{cases} 
      \alpha x & \text{với }x< 0 \\
      x & \text{với }x\geq 0
 \end{cases}
 $$
 trong đó $\alpha$ thường nhận giá trị $0,01$.
 
 Cài đặt trong Python như sau:

```python
def LeakeyReLU(x, alpha=0.01):
    if x < 0:
        return alpha * x
    else:
        return x
```

Dạng sơ đồ, ta có thể hình dung như sau:

```python
import matplotlib.pyplot as plt
import numpy as np


# Leaky Rectified Linear Unit (leaky ReLU) Activation Function
def leaky_ReLU(x):
    data = [max(0.05 * value, value) for value in x]
    return np.array(data, dtype=float)


# Generating data For Graph
x_data = np.linspace(-10, 10, 100)
y_data = leaky_ReLU(x_data)

# Graph
print("\033[1m" + "Fig 6: A Leakey ReLU function curve" + "\033[0m")
plt.plot(x_data, y_data)
plt.legend(["leaky_ReLU"])
plt.grid()
plt.show()
```

#### **1.3.5 Hàm Softmax**
Hàm softmax tổng quát hóa hàm sigmoid và được dùng trong bài toán phân loại nhiều nhóm, trong khi hàm logistic phù hợp với hai nhóm. Biểu diễn là:
$$f(x_i) = \frac{e^{x_i}}{\sum_j e^{x_j}}$$ Mã Python tạo ra phương trình trên là:

```python
def softmax(x):
    return np.exp(x) / np.exp(x).sum(axis=0)
```

$$
f(x) = \begin{cases}

 \end{cases}
 $$
 b) Hàm swish cho bởi $$f(x) = x\sigma(x)$$ trong đó $\sigma(x)$ là hàm sigmoid.

## **2. Lan truyền xuôi**

Ở phần trước, ta nói rằng mạng nơ-ron học bằng cách xếp chồng các nơ-ron thành các lớp. Một mạng nơ-ron được gọi là có $n$ lớp nếu số lớp không tính lớp đầu vào là $n$. Do đó, mạng nơ-ron hai lớp sẽ có một lớp đầu vào $X$, một lớp ẩn $h$ và một lớp đầu ra $y$ như bên dưới:

```python
print("\033[1m" + "Fig 7: A Two Layer Neural Network" + "\033[0m")
layersList = [
    {"title": "input layer", "units": 3, "color": "darkBlue"},
    {"title": "hidden layer", "units": 4},
    {"title": "output layer", "units": 1, "color": "darkBlue"},
]

NNV(layersList).render(save_to_file="my_example3.png")
plt.show()
```

Xét trường hợp có ba biến đầu vào, tức là $x_1$, $x_2$ và $x_3$ như trong sơ đồ trên. Số nơ-ron ở lớp đầu vào do đó là ba, và trong ví dụ này ta cũng giả sử số nơ-ron ở lớp ẩn và lớp đầu ra lần lượt là bốn và một. Như đã nêu, các biến đầu vào sẽ được nhân với các trọng số tương ứng, sau đó ta cộng số hạng chệch trước khi chuyển tổng này đến lớp ẩn, nơi hàm kích hoạt được dùng.

Câu hỏi hiển nhiên ta có thể đặt ra là với các đầu vào đã cho, ta lấy các trọng số tương ứng ở đâu? Trên thực tế, không thể biết đầu vào nào quan trọng hơn trong dự báo và theo tỷ lệ nào. Vì vậy ta gán trọng số và số hạng chệch bằng các số ngẫu nhiên, ký hiệu lần lượt là $W_{xh}$ và $b_h$, trong đó ma trận trọng số có số cột bằng số nơ-ron ở lớp đầu vào và số dòng bằng số nơ-ron ở lớp ẩn để bảo đảm phép nhân ma trận tương thích. Do đó, trong ví dụ trình bày, chiều của $W_{xh}$ là $3\times 4$ và vì vậy $$z_1 = X W_{xh} + b_h.$$
Ta chuyển phương trình này đến lớp ẩn, nơi hàm kích hoạt tác động lên. Nếu hàm kích hoạt trong trường hợp này là hàm sigmoid thì ta có $$a_1 = \sigma (z_1).$$
Đầu ra $a_1$ lại được nhân với một ma trận trọng số mới và cộng với một số hạng chệch mới, rồi chuyển đến lớp đầu ra. Ký hiệu ma trận trọng số và số hạng chệch ở đây lần lượt là $W_{hy}$ và $b_y$, chiều của $W_{hy}$ sẽ là $4\times 1$ vì lớp ẩn có $4$ nơ-ron và lớp đầu ra có một nơ-ron.

Kết quả cho bởi $$z_2 = a_1 W_{hy} + b_y.$$

Cuối cùng, ta áp dụng một hàm kích hoạt nữa lên $z_2$ để có giá trị đầu ra $$\hat{y} = \sigma(z_2).$$ Quá trình trên là cái ta gọi là lan truyền xuôi.

Tóm lại, các bước toán học trong mạng nơ-ron hai lớp như sau:
$$
\begin{cases}
z_1 &= X W_{xh} + b_h\\
a_1 &= \sigma(z_1) \\
z_2 &= a_1 W_{hy} + b_y \\
\hat{y} &= \sigma(z_2)
\end{cases}
$$

Một hàm Python cài đặt các bước trên như bên dưới:

```python
def forward_prop(X, b_h, b_y, W_x_h, W_h_y):
    z_1 = np.dot(X, W_x_h) + b_h
    a_1 = sigmoid(z_1)
    z_2 = np.dot(a_1, W_h_y) + b_y
    y_hat = sigmoid(z_2)

    return y_hat
```

Xong chưa? Có vẻ chưa. Ta phải so sánh đầu ra xấp xỉ với đầu ra thực tế. Vậy làm sao ta biết mô hình hoạt động đạt kỳ vọng? Ta định nghĩa một hàm chi phí/mất mát. Cũng như có nhiều hàm kích hoạt, tồn tại nhiều hàm chi phí. Để xây dựng trực giác ở phần sau, ta dùng hàm chi phí bên dưới
$$J = \frac{1}{n} \sum_{n=1}^n (y_i - \hat{y})^2$$
tổng bình phương chênh lệch giữa đầu ra thực tế và đầu ra dự báo, trong đó $n$ là số biến dùng trong huấn luyện.

Bây giờ ta đã học cách dự báo đầu ra của mạng nơ-ron, ở phần tiếp theo ta sẽ tinh chỉnh điều này để có xấp xỉ tốt hơn.

## **3. Lan truyền ngược**

Lúc này, câu hỏi còn vương trong đầu ta có thể là mạng nơ-ron học dự báo đúng đầu ra bằng cách nào? Hàm chi phí đóng vai trò gì trong tất cả những điều này? Câu trả lời ngắn là ta muốn hàm chi phí càng nhỏ càng tốt, và do đó ở mỗi vòng lặp ta sẽ tìm cách cực tiểu hóa hàm chi phí. Nhưng tại sao? Nếu hàm chi phí lớn thì nghĩa là chênh lệch giữa đầu ra dự báo và thực tế cao.

Cách duy nhất để cực tiểu hóa hàm chi phí là thay đổi một số tham số, nhưng ta không thể thay đổi giá trị đầu vào và đầu ra. Do đó ta chỉ có thể thay đổi các trọng số và số hạng chệch vì ban đầu chúng được gán bằng số ngẫu nhiên. Vai trò của lan truyền ngược là giúp mạng nơ-ron học cách cập nhật các trọng số sao cho cực tiểu hóa hàm chi phí, qua đó tăng độ chính xác của dự báo. Ta dùng phương pháp gọi là giảm gradient để cập nhật các trọng số.

Giảm gradient là thuật toán tối ưu hóa thường dùng trong học máy để tìm giá trị tối ưu của các trọng số sao cho cực tiểu hóa hàm chi phí hay sai số.

Để hiểu cách giảm gradient tìm trọng số tối ưu, ta giải thích bằng trực giác. Từ tên "giảm gradient", hãy nghĩ đến việc đi xuống một con dốc. Kỹ thuật này cố tìm một tổ hợp tham số trên một đường cong để bảo đảm đầu ra cho giá trị chi phí thấp.

Vì ta quan tâm đến việc tìm trọng số tối ưu, hãy xét đường cong bên dưới như một hàm của $w$, tức là $E(w)$ như bên dưới:

```python
import matplotlib.pyplot as plt
import numpy as np

w = np.linspace(-3, 3, num=100)

Ew = []
for i in range(len(w)):
    Ew.append(w[i] ** 4 - 5 * w[i] ** 2 - 3 * w[i])

print("\033[1m" + "Fig 8: A Non-convex Curve" + "\033[0m")
plt.plot(w, Ew)
plt.grid()
plt.axvline()
plt.axhline()
plt.show()
```

Lý tưởng nhất là ta chạm đến cực tiểu toàn cục (đáy thấp nhất của đường cong), nhưng giảm gradient không nhất thiết đến được điểm này, dù sẽ đến gần. Trước hết ta chọn các giá trị ban đầu của $w$. Giảm gradient sẽ cố đạt điểm cực tiểu cục bộ, và vì nó chỉ đi xuống chứ không đi lên, bằng cách tăng hoặc giảm giá trị $w$ dần theo các bước nhỏ, ta sẽ thấy nó đi xuống dốc về phía cực tiểu toàn cục. Ta gọi các bước này là tốc độ học, ký hiệu là $\alpha$.

Lặp lại quá trình này qua nhiều vòng lặp, giảm gradient sẽ cho ta các trọng số tối ưu bảo đảm chi phí ở mức cực tiểu. Hãy nhớ rằng ở lan truyền xuôi ta dừng ở lớp đầu ra. Ở lan truyền ngược, ta đi theo thứ tự đảo ngược, tức là từ lớp đầu ra về lớp đầu vào, đồng thời tính gradient của hàm chi phí để cực tiểu hóa sai số. Sau khi tính xong các gradient, ta có thể cập nhật các trọng số như sau: $$W_{new} = W - \alpha \frac{\partial J}{\partial W}$$
trong đó $\alpha$ là tốc độ học.

Tốc độ học định hướng loại bước đi, tức là tốc độ học nhỏ nghĩa là ta đi một bước nhỏ xuống dốc và do đó quá trình giảm gradient trở nên chậm. Điều ngược lại cũng đúng như trong các sơ đồ bên dưới.

**Hình 9**: Giảm gradient với các tốc độ học khác nhau

<img src="https://www.jeremyjordan.me/content/images/2018/02/Screen-Shot-2018-02-24-at-11.47.09-AM.png" alt="Giảm gradient với bước lớn" width="1000"/> 

##### Nguồn: [Jeremy Jordan, Setting the Learning Rate of Your Neural Network, 2018](https://www.jeremyjordan.me/nn-learning-rate/)

Toàn bộ quy trình đi từ lớp đầu ra về lớp đầu vào trong khi cập nhật các trọng số bằng một thuật toán tối ưu này gọi là lan truyền ngược.

Trong ví dụ mạng nơ-ron hai lớp ta xét ở phần trước, ta có hai trọng số, $W_{xh}$ đưa đầu vào vào lớp ẩn và $W_{hy}$ đưa từ lớp ẩn vào lớp đầu ra. Để tìm các trọng số tối ưu cho sai số nhỏ nhất, ta sẽ tìm đạo hàm của $J$ theo các trọng số.

Ta bắt đầu với $W_{hy}$ vì ta đi ngược, từ lớp đầu ra về lớp đầu vào. Nhớ lại hàm chi phí $J$ cho bởi:
$$J = \frac{1}{n}\sum_{i = 1}^{n} (y_i - \hat{y})^2$$
và $\hat{y} = \sigma(z_2)$ trong đó $z_2 = a_1 W_{hy} + b_y$. Để tìm đạo hàm của $J$ theo $W_{hy}$ ta dùng quy tắc chuỗi, tức là
$$\frac{\partial J}{\partial W_{hy}} = \frac{\partial J}{\partial \hat{y}} \times \frac{\partial \hat{y}}{\partial z_2} \times \frac{\partial z_2}{\partial W_{hy}} \hspace{10mm} (1)$$

Lưu ý:
$$\frac{\partial J}{\partial \hat{y}} = (y - \hat{y})$$
$$\frac{\partial \hat{y}}{\partial z_2} = \sigma ' (z_2)$$
trong đó $\sigma '$ là đạo hàm của hàm kích hoạt sigmoid.

Đạo hàm của hàm sigmoid $\sigma (z) = \frac{1}{1 + e^{-z}}$ là
$$\sigma ' (z) = \frac{e^{-z}}{(1+e^{-z})^2}$$
Cuối cùng, $$\frac{\partial z_2}{\partial W_{hy}} = a_1$$
Kết hợp các đạo hàm vào phương trình (1) ở trên, ta được
$$\frac{\partial J}{\partial W_{hy}} = (y - \hat{y}) \times \sigma ' (z_2) \times a_1 \hspace{10mm} (2)$$

Tiếp theo, ta tìm đạo hàm của $J$ theo trọng số $W_{xh}$. Nhớ lại:
$$\hat{y} = \sigma(z_2)$$
$$z_2 = a_1 W_{hy} + b_y$$
$$a_1 = \sigma(z_1)$$
$$z_1 = X W_{xh} + b$$
Theo quy tắc chuỗi, ta được
$$\frac{\partial J}{\partial W_{xh}} = \frac{\partial J}{\partial \hat{y}} \times \frac{\partial \hat{y}}{\partial z_2} \times \frac{\partial z_2}{\partial a_1} \times \frac{\partial a_1}{\partial z_1} \times \frac{\partial z_1}{\partial W_{xh}}\hspace{10mm} (3)$$
trong đó:
$$\frac{\partial z_2}{\partial a_1} = W_{hy}$$
$$\frac{\partial a_1}{\partial z_1} = \sigma ' (z_1)$$
$$\frac{\partial z_1}{\partial W_{xh}} = X$$

Do đó:
$$\frac{\partial J}{\partial W_{xh}} = (y - \hat{y}) \times \sigma ' (z_2) \times W_{hy} \times \sigma ' (z_1) \times X \hspace{10mm} (4)$$

Vì bây giờ ta đã có các gradient, ta có thể cập nhật các trọng số như sau:
$$W_{hy} = W_{hy} - \alpha \frac{\partial J}{\partial W_{hy}} \hspace{10mm} (5)$$
$$W_{xh} = W_{xh} - \alpha \frac{\partial J}{\partial W_{xh}} \hspace{10mm} (6)$$

Bây giờ ta cài đặt lan truyền ngược trong Python. Ta định nghĩa $d Why$ và $d Whx$ lần lượt đại diện cho $\frac{\partial J}{\partial W_{hy}}$ và $\frac{\partial J}{\partial W_{xh}}$.

Trước hết ta viết một hàm cho đạo hàm của hàm sigmoid.

```python
def sigmoid_derivative(z):
    return np.exp(-z) / ((1 + np.exp(-z)) ** 2)
```

Định nghĩa $(y - \hat{y}) \sigma ' (z_2)$ là:

`delta2 = np.multiply((y_hat - y)`, `sigmoid_derivative(z_2))`

Do đó gradient theo $W_{hy}$ là:

`dWhy = np.dot(a1.T, delta2)`

và theo $W_{xh}$ là:

`delta1 = np.dot(delta2, Why.T) * sigmoid_derivative(z1)`

`dWxh = np.dot(X.T, delta1)`

Ta cập nhật các trọng số tương ứng:

`Wxh = Wxh - alpha * dWxh`

`Why = Why - alpha * dWhy`

Vậy hàm trở thành:

```python
def backward_prop(y_hat, z_1, a_1, z_2, alpha, W_x_h, W_h_y, X):
    delta2 = np.multiply((y_hat - y), sigmoid_derivative(z_2))
    d_W_h_y = np.dot(a_1.T, delta2)

    delta1 = np.dot(delta2, W_h_y.T) * sigmoid_derivative(z_1)
    d_W_x_h = np.dot(X.T, delta1)

    W_x_h = W_x_h - alpha * d_W_x_h
    W_h_y = W_h_y - alpha * d_W_h_y

    return W_x_h, W_h_y
```
