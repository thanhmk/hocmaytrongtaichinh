## **1. Cơ bản về mạng nơ-ron**

Mạng nơ-ron (NN) thường được mô tả là cách mô phỏng bộ não con người bằng máy tính. Khi nghĩ về quá trình nhận diện một hình ảnh, chẳng hạn hình ảnh con chó, trước hết ta nhìn hình bằng mắt, qua đó gửi đầu vào đến não. Bước tiếp theo là não phân tích hình ảnh dựa trên những gì ta đã thấy về chó trong quá khứ. Cuối cùng, não đưa ra quyết định/kết luận rằng đây là hình ảnh con chó. Tóm tắt quá trình này, ta có:

Đầu vào -> phân tích -> quyết định.

Việc này khá đơn giản với chúng ta; tuy nhiên để máy tính thực hiện cùng nhiệm vụ nhận diện hình ảnh con chó thì không hề đơn giản. Quá trình tương tự ở chỗ máy tính cũng thực hiện một quá trình giống vậy:

Đầu vào -> xử lý/tính toán -> đầu ra.

Ta xét một ví dụ để hiểu cách mạng nơ-ron khám phá mối quan hệ giữa đầu vào và đầu ra. Ta xét bài toán đơn giản là đổi khối lượng từ pound (lbs) sang kilogram (kg). Giả sử ta không biết công thức quy đổi và chỉ có quan sát sau:

|  |  |
|:---|:---|
|**kg** |  **lbs** |
|50 | 110,23  |


---

Ở đây ta đưa ra một giả định quan trọng: mối quan hệ giữa kg và lbs là tuyến tính. Mình biết đây là phỏng đoán tốt vì ta biết công thức là $lbs = 2,2046 \times kg$, nhưng thường thì ta không biết mối quan hệ. Vì vậy ta đưa ra phỏng đoán rằng mối quan hệ là lbs = $c\times$kg, với $c$ là hệ số góc. Phương trình tuyến tính này chính là cái gọi là *hàm kích hoạt* (activation function) trong mạng nơ-ron. Ta bắt đầu bằng phỏng đoán $c=1,5$, và với phỏng đoán này, ước lượng lbs là lbs = 1,5(50) = 75 lbs. Như vậy ta có vẻ đã ước lượng thấp đi 110,23 - 75 = 35,23 lbs. Được rồi, ta đoán lại, tăng $c$ lên dựa trên phản hồi từ phỏng đoán đầu, và chọn $c=3$, cho ước lượng thứ hai là 50(3) = 150 lbs. Lần này có vẻ ta ước lượng cao hơn 39,77 lbs. Do đó ta cần chọn giá trị $1,5 < c < 3 $. Ta làm lặp đi lặp lại như vậy đến khi đạt giá trị $c=2,2$, cho kết quả 110 lbs. Khá gần. Ta có thể quyết định thế là đủ gần, và dừng ở $c=2,2$. Ta có thể tóm tắt các vòng lặp bên dưới,

|  |  |  |  |
|:---|:---|:---|:---|
|**vòng lặp** |  $c$ |  ước lượng lbs|  (Thực tế - dự báo) |

Cột cuối cho ta biết phỏng đoán lệch bao xa so với thực tế. Đây là một ví dụ về *hàm mất mát* (loss function) trong mạng nơ-ron. Độ lớn của chênh lệch giữa thực tế và ước lượng cho biết ta cần điều chỉnh phỏng đoán tiếp theo cho $c$ như thế nào. Hãy hiểu rằng chênh lệch lớn thì phỏng đoán $c$ tiếp theo sẽ khác nhiều so với lần trước. Loại phản hồi này là cái ta gọi là *bộ tối ưu* (optimizer) trong mạng nơ-ron. Với ví dụ này, ít nhiều ta đã nắm được cách hoạt động hay luồng của một mạng nơ-ron. Có thể tóm tắt ở hình 1 bên dưới. Bạn sẽ thấy một thành phần 'trọng số' trong sơ đồ, mà ta sẽ nói đến ở phần sau. Để làm vậy, ta sẽ xem kiến trúc của mạng nơ-ron.

## **3. Lan truyền xuôi và lan truyền ngược**

Mạng nơ-ron đạt được dự báo qua một quá trình gọi là lan truyền. Ở mỗi bước xuôi có các phép nhân ma trận diễn ra. Máy tính làm các phép tính này tương đối nhanh, và cần hiểu quá trình đằng sau nó. Để bắt đầu, trước hết ta phải xác định chiều của đầu vào ở lúc đầu, tức là ở lớp đầu vào. Ta đã nói ở trên rằng số nút ở lớp đầu vào tương ứng với số biến dự báo hay đặc trưng. Đây là số cột của ma trận đầu vào, và số dòng tương ứng với số quan sát, giả sử là $N$. Do đó ta có ma trận đầu vào $N\times3$ cho hình 2. Tổng quát, ký hiệu số đặc trưng là $p$ nên hình 2 có $p=3$. Ta ký hiệu các lớp như sau:


1.   Lớp đầu vào: $X_0$
2.   Lớp ẩn: $X_1$
3.   Lớp đầu ra: $Y$

Ta sẽ chạm đến một chút đại số tuyến tính nhưng không có gì nặng ngoài phép nhân ma trận. Ở bước lan truyền xuôi đi từ $X_0$ đến $X_1$, có các phép tính sau diễn ra, có thể biểu diễn bằng phép nhân ma trận,

$$
\begin{equation} 
X_{1} = f(X_{0} w_{0}+b_{0}),
\tag{1} 
\end{equation}
$$
trong đó $w_0$ là ma trận $p\times4$ các *trọng số* vì lớp ẩn có 4 nút, và $f$ là hàm kích hoạt do người dùng chỉ định. Số hạng $b$ là số hạng chệch giúp hàm kích hoạt dịch chuyển, cho phép thuật toán linh hoạt hơn khi khớp dữ liệu. Hàm kích hoạt, ngoài việc quy định tính tuyến tính/phi tuyến của dữ liệu, còn quyết định nút nào được kích hoạt bằng cách kiểm soát giá trị của các trọng số. Chẳng hạn, giả sử Tuổi không phải biến dự báo mạnh cho việc vỡ nợ khoản vay thì trọng số của nút đó sẽ nhỏ. Để đi tuần tự đến lớp đầu ra, ta có phép tính sau,

$$
\begin{equation} 
Y = f(X_{1} w_{1}+b_{1})=f(f(X_{0} w_0 +b_{0}) w_1+b_{1}),
\tag{2} 
\end{equation}
$$

bằng cách thay (1) vào $X_1$. Chiều của $w_1$ sẽ là 4x2 vì lớp ẩn có 4 nút và lớp đầu ra có 2 nút. Quá trình nhân ma trận tuần tự này làm các chiều thay đổi như sau:


*   $X_{0}\times w_0$: $(N\times p) \times (p\times 4) = N\times 4$ (hãy nghĩ $p$ triệt tiêu lẫn nhau)
*   $X_{1}\times w_1$: $(N\times 4) \times (4\times 2) = N\times 2$

Các chiều của trọng số và các lớp là cần thiết khi chỉ định kiến trúc của mạng nơ-ron trong một ngôn ngữ lập trình như Python.

Đến giờ, ta đã cho thấy quá trình lan truyền xuôi để có đầu ra; tuy nhiên chính quá trình lan truyền ngược mới giúp mạng nơ-ron học và cải thiện. Đây là quá trình phản hồi từ hàm mất mát đến bộ tối ưu, cuối cùng cho thuật toán biết cách điều chỉnh các trọng số. Rốt cuộc, chính các trọng số và một số hạng chệch sẽ được điều chỉnh để đạt hiệu suất tối ưu. Một lợi ích lớn của mạng nơ-ron là ta không phải giới hạn ở một lớp ẩn, và bằng cách tăng số lớp, ta tăng cơ hội giải quyết các quan hệ dữ liệu phi tuyến. Có một vấn đề mà mạng nơ-ron gặp phải khi tăng số lớp, gọi là *vấn đề gradient biến mất* (vanishing gradient).

## **4. Gradient biến mất**

Ta đã nhắc ngắn gọn vai trò của hàm kích hoạt ở mục 1. Có nhiều hàm kích hoạt để chọn, mỗi hàm có lợi ích riêng. Bạn đọc xem (Nwankpa 7) để có danh sách ngắn các hàm này. Hàm kích hoạt sigmoid là hàm phổ biến dùng trong bài toán phân loại nhị phân. Nó nén đầu ra vào một khoảng nhỏ giữa 0 và 1. Dạng hàm có thể quen nếu bạn từng làm việc với mô hình hồi quy logistic, vì nó được tính như sau,

$$
\begin{equation}
σ(x) = \frac{1}{1+e^{-x}}
\end{equation}
$$

Điều này cho ta biết rằng với một thay đổi lớn của $x$ hay đầu vào, ta chỉ có thay đổi nhỏ ở đầu ra. Nhưng ta biết gradient được cho bởi,

$$
\begin{equation}
\text{Gradient} = \frac{\text{thay đổi ở đầu ra}}{\text{thay đổi ở đầu vào}}
\end{equation}
$$

do đó điều này hàm ý gradient có xu hướng nhận giá trị nhỏ với hàm sigmoid. Mối liên hệ của hàm sigmoid với hàm mất mát, giả sử $L$, là ta biết đầu ra phụ thuộc vào hàm kích hoạt. Mạng càng có nhiều lớp thì quá trình càng cần nhiều hàm kích hoạt. Ý tưởng của lan truyền ngược ở bảng 1 là cực tiểu hóa $L$ hay chênh lệch ở cột cuối thông qua phản hồi. Giải tích rất tiện khi ta muốn cực tiểu hóa một hàm $f$ theo đầu vào $x$. Quá trình gồm việc xác định giá trị $x$, giả sử $x'$, làm đạo hàm hay gradient của $f$ = 0. Do đó lan truyền ngược bao gồm việc tìm đạo hàm của lớp đầu tiên bằng cách nhân các đạo hàm của từng lớp đi từ lớp cuối đến lớp đầu, một quá trình gọi là quy tắc chuỗi. Vì vậy, vấn đề xuất hiện khi tích này tiến về không khi có nhiều lớp, nhất là với hàm kích hoạt sigmoid. Điều này ngăn việc cập nhật tối ưu các trọng số và số hạng chệch.

Từ năm 2009 đã có những cải tiến cho mạng nơ-ron để khắc phục vấn đề này, như các hàm kích hoạt tốt hơn như ReLu, các cách khởi tạo trọng số tốt hơn, và các cách tối ưu hóa được cải thiện. Mạng nơ-ron đã đi một chặng đường dài kể từ khi ra đời, và với sự sẵn có của GPU, tốc độ huấn luyện các thuật toán này đã được cải thiện rất nhiều.
