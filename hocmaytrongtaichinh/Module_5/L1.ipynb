## **1. Giới thiệu**

Máy vector hỗ trợ là một cách tiếp cận để giải bài toán phân loại và ngày càng phổ biến từ khi ra đời vào những năm 1990. Giống hồi quy logistic, nó hiệu quả với dữ liệu phân tách tuyến tính được, tức là dữ liệu có thể tách bằng một ranh giới tuyến tính. Tuy nhiên có những siêu tham số có thể dùng để điều chỉnh ranh giới này nhằm xử lý dữ liệu phi tuyến. Ranh giới này là cái ta gọi là siêu phẳng phân tách.

## **2. Siêu phẳng phân tách**

Siêu phẳng đơn giản là một mặt phẳng tuyến tính trong không gian $p$ chiều bất kỳ. Ví dụ là đường thẳng trong không gian 2 chiều như thấy ở hình 1, tức là với $p=2$ chiều, siêu phẳng là một không gian con phẳng $p-1=1$ chiều. Ở hình 1 bên dưới, đường xanh dương là một siêu phẳng trong không gian hai chiều.

**Hình 1:** Ranh giới tuyến tính phân tách dữ liệu có nhóm mục tiêu nhị phân

![](https://raw.githubusercontent.com/thanhmk/hocmaytrongtaichinh/main/Module_5/CustomLinSepData.png)


Về mặt toán học, trong ví dụ này siêu phẳng có thể biểu diễn bằng phương trình đường thẳng,
$
\begin{equation} 
\beta_{0} + \beta_{1}X_{1} + \beta_{2}X_{2}= 0,
\tag{1} 
\end{equation}
$
do đó, tổng quát cho $p$ chiều,

$$
\begin{equation} 
\beta_{0} + \beta_{1}X_{1} + \beta_{2}X_{2} + \cdots +\beta_{p}X_{p}= 0, 
\tag{2}
\end{equation}
$$
hay,

$$
\begin{equation} 
\sum_{j=1}^{p}\beta_{j}X_{j} + \beta_{0} = 0. 
\tag{3}
\end{equation}
$$

Giả sử phương trình của siêu phẳng phân tách biểu diễn ở hình 1 bằng đường xanh dương là,

$$
\begin{equation}
-4 + 4X_{1} + X_{2}= 0. 
\tag{4}
\end{equation}
$$

Bây giờ nếu có một quan sát với tọa độ $(X_{1}, X_{2})$ là $(6, -10)$, thì thay điểm này vào (4) ta được,
$$
\begin{equation} 
-4 + 4(6) + (-10)= 10 > 0. 
\end{equation}
$$

Quan sát này nằm phía trên siêu phẳng, cùng với nhóm 1 (các tam giác đỏ). Quy tắc phân loại nhóm mục tiêu cho một quan sát $(X_{1}, X_{2})$ bất kỳ sẽ là,

$$
\begin{equation} 
-4 + 4X_{1} + X_{2} > 0 \text{ thì Nhóm 1}, 
\end{equation}
$$
hoặc

$$
\begin{equation} 
-4 + 4X_{1} + X_{2} < 0 \text{ thì Nhóm 0}. 
\end{equation}
$$

Câu hỏi bạn có thể đặt ra là làm gì với một quan sát nằm trên ranh giới, tức là

$$
\begin{equation} 
-4 + 4X_{1} + X_{2} = 0 \text{ ?} 
\end{equation}
$$

Để giải quyết, ta cần xem siêu phẳng này được suy ra thế nào, và để làm vậy ta xem Bộ phân loại biên cực đại.

## **3. Bộ phân loại biên cực đại**


Bạn nên xem hình 2 để hiểu phần này. Trong ví dụ ở hình 2, bộ phân loại tối ưu cho hai biến dự báo là đường xanh dương. Siêu phẳng tối ưu này được dựng sao cho có các "vùng đệm" cách đều ở hai bên, thể hiện bằng các đường nét đứt. Khoảng cách vuông góc giữa các đường nét đứt gọi là biên (margin) và siêu phẳng tối ưu đạt được khi biên này lớn nhất. Một cách nghĩ khác là bộ phân loại biên cực đại làm cực đại khoảng cách đến điểm dữ liệu gần nhất. Bạn sẽ thấy các đường nét đứt chạm vào dữ liệu ở cả hai phía, nhưng chỉ cần chạm vào các điểm dữ liệu ở một phía là đã đạt biên. Các vùng đệm thể hiện bằng đường nét đứt này gọi là các vector hỗ trợ.

**Hình 2:** Ranh giới tối ưu phân tách hai nhóm

![](https://raw.githubusercontent.com/thanhmk/hocmaytrongtaichinh/main/Module_5/svm_annotate_scrs.png)


Với bài toán phân loại nhị phân này cho các biến dự báo $X_{1}$ và $X_{2}$, ranh giới tối ưu có được bằng cách cực đại hóa biên $M$, với điều kiện:

$$ 
\begin{equation} 
\sum_{j=1}^{2}\beta_{j}^{2} = 1.
\end{equation}
$$

Với số chiều biến dự báo cao hơn, giả sử $p$, điều này mở rộng thành ràng buộc,

$$
\begin{equation}  
\sum_{j=1}^{p}\beta_{j}^{2} = 1.
\tag{4}
\end{equation}
$$

Ràng buộc ở phương trình (4) nhằm bảo đảm ta xét khoảng cách vuông góc từ điểm dữ liệu thứ $j$ đến siêu phẳng.

Thay vì nhóm 0 và 1 cho bài toán nhị phân, giả sử nhãn nhóm là -1 và 1, tức là $y_i$ là -1 hoặc 1 cho quan sát thứ i. Từ (3), một phân loại đúng sẽ là

$$
\begin{equation}
y_{i}(\sum_{j=1}^{p}\beta_{j}X_{ij} + \beta_{0}) > 0,
\tag{5} 
\end{equation}
$$

và do đó các phân loại sai là

$$
\begin{equation}
y_{i}(\sum_{j=1}^{p}\beta_{j}X_{ij} + \beta_{0}) \leq 0.
\tag{6} 
\end{equation}
$$

Một phương pháp gọi là thuật toán Perceptron xác định các trọng số $\beta_{j}$ bằng cách xét các phân loại sai trong (6) và cập nhật trọng số theo cách lặp. Chi tiết thuật toán này sẽ không được trình bày trong khóa học.

Để an toàn hơn, hãy nghĩ khoảng cách vuông góc từ siêu phẳng đến vector hỗ trợ là $D$. Với lớp bảo vệ thêm này, ta có thể đổi (5) thành

$$
\begin{equation} 
y_{i}(\sum_{j=1}^{p}\beta_{j}X_{ij} + \beta_{0}) > D. 
\end{equation}
$$

Một điều cần lưu ý là SVM không cho ra xác suất của nhóm mục tiêu; tuy nhiên có một số phương pháp để thu được điều này, như hiệu chỉnh Platt và hồi quy đẳng điệu (isotonic regression). Đến giờ ta mới xét dữ liệu phân tách tuyến tính được, nhưng trong thực tế thường không như vậy. Thuật toán perceptron không hoạt động tốt với dữ liệu không phân tách tuyến tính được. Điều này đưa ta đến biên cứng và biên mềm.

## **4. Các loại biên**

Biên cứng là khi không có phân loại sai nào xảy ra hoặc được cho phép. Biên mềm là khi có thể chỉ định số lần phân loại sai. Một siêu tham số gọi là tham số điều chuẩn hóa, $c$, xác định mức độ quan trọng của việc có phân loại sai. $c$ càng cao thì mức phạt càng lớn, tức là rất quan trọng việc không có phân loại sai. Khi tăng $c$, ta tiến về biên cứng, tức là không có phân loại sai. Vấn đề của việc dùng siêu phẳng phân tách hoàn hảo các nhóm trong tập huấn luyện là nó có thể không đạt hiệu suất tương tự với dữ liệu chưa thấy hay dữ liệu kiểm tra. Đây là vấn đề quá khớp. Khi tăng $c$, thực ra ta đang theo cách tiếp cận thận trọng hơn bằng cách dùng biên rất nhỏ, tức là các đường nét đứt ở hình 2 gần siêu phẳng hơn nhiều (đường xanh dương liền). Một $c$ nhỏ hơn đẩy các biên ra xa hơn vì ta cho phép bộ phân loại chấp nhận nhiều phân loại sai hơn. Nói cách khác, $c$ nhỏ tạo ra biên mềm, đổi lại có một số phân loại sai. Biến lỏng (slack variable) đo khoảng cách của các điểm phân loại sai đến biên, tức là những điểm không nằm hoàn toàn ở đúng phía của biên. Tham số $c$ là ràng buộc lên các biến lỏng. Cần lưu ý một số sách dùng $c$ theo chiều ngược lại, định nghĩa nó là cận trên của tổng mọi biến lỏng. Ta dùng ký hiệu $c$ để khớp với cách dùng trong python. Xem https://stats.stackexchange.com/q/488107 để biết cách diễn giải toán học. Dữ liệu mới đưa vào tập huấn luyện có thể làm đổi siêu phẳng; tuy nhiên cần lưu ý chỉ những điểm nằm trên biên và ở sai phía của siêu phẳng mới gây ra thay đổi. Tính chất này nhấn mạnh sự không nhạy với giá trị ngoại lai. Siêu tham số gamma có thể được điều chỉnh để các điểm ở gần hay xa ranh giới quyết định có ảnh hưởng. Gamma thấp lấy các điểm ở xa và ngược lại. Cuối cùng ta đến bài toán dữ liệu phi tuyến và cách SVM vượt qua nó bằng các nhân (kernel).

## **5. Nhân (Kernel)**


Giả sử các nhóm mục tiêu phân bố trong không gian đặc trưng giữa hai biến dự báo như ở biểu đồ bên trái của hình 3.

**Hình 3:** Trái: dữ liệu không phân tách tuyến tính được. Phải: dữ liệu được chiếu lên thêm một chiều $Z$ sao cho $Z = X1^2 + X2^2$.


![](https://raw.githubusercontent.com/thanhmk/hocmaytrongtaichinh/main/Module_5/KernelTransformData.png)


Dùng cách tiếp cận biên cứng, một đường thẳng sẽ không chia các nhóm cho phù hợp. Một cách làm là thêm một biến dự báo là hàm của các biến gốc sao cho

$$
\begin{equation} 
Z = X1^2 + X2^2.
\end{equation}
$$

Chiếu không gian đặc trưng lên thêm một chiều sẽ cho thấy phân phối của các nhóm mục tiêu ở biểu đồ bên phải của hình 3. Bây giờ ta thấy một mặt phẳng có thể phân tách hiệu quả các nhóm mục tiêu. Việc ánh xạ lên các chiều cao hơn này do các nhân (kernel) thực hiện. Thuật toán SVM có nhiều nhân để chọn, như nhân đa thức và hàm cơ sở xuyên tâm Gauss (RBF), kể một vài loại. Để có danh sách đầy đủ hơn, xem https://data-flair.training/blogs/svm-kernel-functions/. Một điều cần lưu ý là ví dụ minh họa này ban đầu chỉ có hai biến dự báo; do đó với nhiều biến dự báo hơn, phép biến đổi tối ưu không chỉ khó tìm mà còn khó tính. Các nhân đạt được điều này theo cách tiết kiệm bộ nhớ hơn. Toán học chi tiết của nhân không thuộc phạm vi module này, nhưng bạn đọc quan tâm xem https://www.analyticsvidhya.com/blog/2020/10/the-mathematics-behind-svm/. Điều ta cần hiểu là nghiệm của việc tính siêu phẳng tối ưu phụ thuộc vào tích vô hướng (tích trong) của các quan sát. Tích trong của hai vector $a = (a_{1}, a_{2},⋯,a_{n})$ và $b = (b_{1}, b{2},⋯,b_{n})$ có cùng độ dài $n$ là

$$
\begin{equation}
\left<a,b\right> = \sum_{i=1}^{n}a_i b_i.
\tag{7}
\end{equation}
$$

Từ (7), bạn có thể nhận thấy đây "phần nào" là cách ta tính $Z$ trong ví dụ trên, tức là,

$$
\begin{align}
Z = \sum_{j=1}^{2} \left<Xj_{i},Xj_{i'}\right>
= K(X_{i},X_{i'})
\end{align}
$$

cho hai quan sát khác nhau $i$ và $i'$. Ta dùng $K$ để ký hiệu nhân.

Tổng quát, một vector hỗ trợ tuyến tính có thể biểu diễn là

$$
\begin{equation}
y(x) = \beta_0 + \sum_{i}^{n}\alpha_{i}\left<X, X_{i} \right>
\end{equation}
$$

cho một điểm mới $X$, trong đó các tham số $\alpha_{i}$ là ước lượng trên mỗi quan sát $i$. Điều này nghĩa là cần tính tích trong giữa điểm mới $X$ và mọi quan sát khác để ước lượng $\alpha$; tuy nhiên $\alpha$ bằng không với mọi quan sát không phải vector hỗ trợ. Do đó ta thấy việc xác định các $\alpha$ đỡ tốn tính toán hơn.

Nhân Gauss RBF có công thức khác như bên dưới,

$$
\begin{align}
K(X_{i},X_{i'}) = \text{exp} \left(-\gamma ||X_{i}-X_{i'}||^2\right)
\end{align}
$$

trong đó $||X_{i}-X_{i'}||$ là khoảng cách Euclid giữa quan sát $X_i$ và $X_{i'}$. Công thức này có thể quen thuộc nếu bạn đã thấy công thức của phương trình Gauss $\text{exp} \left(-\gamma ||X-\mu||^2\right)$ với trung bình $\mu$. Tham số $\gamma$ tương đương với độ rộng của phân phối Gauss hay độ lệch chuẩn $\sigma$, tức là $\gamma = 1/2\sigma^2$. Do đó đây là siêu tham số do người dùng cung cấp và $\gamma$ quá lớn có thể dẫn đến quá khớp. Nhân Gauss thường được dùng khi dữ liệu gốc không phân tách tuyến tính được và đôi khi là nhân mặc định trong các ngôn ngữ lập trình. Tốt nhất là biết nhân mặc định dùng trong mô hình SVM và thử nghiệm với các nhân khác nhau.
