## **1. Giới thiệu**

Trong học có giám sát, đầu ra (biến mục tiêu) mà ta muốn dự báo đã được biết trước. Điều đó cho phép ta đánh giá chất lượng mô hình bằng các thước đo như recall, độ chính xác, sai số bình phương trung bình, v.v. Trong học không giám sát, biến mục tiêu không có sẵn. Mô hình không giám sát giúp tìm ra các mẫu hình trong dữ liệu đã được gom nhóm mà không có phản hồi nào đi kèm.

Bảng dưới đây liệt kê một số khác biệt giữa mô hình không giám sát và mô hình có giám sát.

## **2. Phân cụm**

Phân cụm là quá trình tổ chức các điểm dữ liệu thành các nhóm sao cho các thành viên trong mỗi nhóm có một số thuộc tính tương tự nhau. Trong biểu đồ dưới đây, ta có thể nói các phần tử trong Cụm 0 giống nhau nhưng khác các phần tử ở cụm khác, chẳng hạn Cụm 1.

### **2.1 Phân cụm dựa trên khoảng cách**

Với một tập các điểm dữ liệu, ta có thể dùng khoảng cách giữa các điểm để gom chúng thành các cụm sao cho
- Khoảng cách bên trong giữa các điểm riêng lẻ (trong cụm) phải nhỏ, tức các điểm của một cụm phải gần nhau.
- Khoảng cách bên ngoài giữa các điểm (giữa các cụm) phải lớn, tức các điểm ở những cụm khác nhau phải khác biệt.

### **2.2 Mục tiêu của phân cụm**

Với một tập dữ liệu không có nhãn, ta áp dụng phân cụm để tìm cấu trúc và các nhóm tồn tại trong dữ liệu. Không có tiêu chí hoàn hảo nào cho việc phân cụm, và người dùng phải tự đưa ra tiêu chí để đảm bảo kết quả phù hợp với nhu cầu của mình.

Với bất kỳ phương pháp phân cụm nào được chọn, ta cần định nghĩa một **thước đo tương đồng** phù hợp để dùng khi tìm các cụm.

### **2.3 Các thước đo độ gần**

Khi phân cụm, ta phải định nghĩa một thước đo độ gần giữa hai điểm, trong đó độ gần là mức tương đồng hoặc khác biệt của dữ liệu đối với nhau. Thước đo tương đồng $S(x_i, x_k)$ lớn khi các điểm $x_i, x_k$ giống nhau, còn thước đo khác biệt $D(x_i x_k)$ nhỏ khi các điểm $x_i, x_k$ giống nhau.

Dưới đây là các thước đo tương đồng ta có thể dùng:

1. Vector: Khoảng cách cosine
$$s(\vec{x}, x') = \frac{x^t x'}{||x||||x'||}$$

2. Tập hợp: Khoảng cách Jaccard
$$J(A, B) = \frac{|A\cap B|}{|A\cup B|} = \frac{|A\cap B|}{|A| + |B| - |A\cap B|}$$
Nếu hai tập $A$ và $B$ đều rỗng thì ta định nghĩa $J(A, B)=1$, ngược lại $0\leq J(A, B) \leq 1$

3. Điểm: Khoảng cách Euclid
$$d(x, x') = \left(\sum_{k=1}^p \left|x_k - x'_k\right|^q\right)^{1/q}$$


### **2.4 Các thuật toán phân cụm**

Các thuật toán phân cụm có thể chia thành bốn loại như sau:
1. Phân cụm loại trừ
2. Phân cụm chồng lấn
3. Phân cụm phân cấp
4. Phân cụm xác suất

Trong phân cụm loại trừ, dữ liệu được gom nhóm sao cho một điểm thuộc về một cụm thì không thể thuộc cụm khác.

Phân cụm chồng lấn dùng tập mờ để gom dữ liệu sao cho một điểm có thể thuộc nhiều hơn một cụm nhưng với mức độ thuộc khác nhau. Do đó ta gom dữ liệu theo một giá trị mức độ thuộc phù hợp.

Trong phân cụm phân cấp, mỗi điểm dữ liệu ban đầu được coi là một cụm, và ở mỗi vòng lặp ta tìm hợp của hai cụm gần nhau nhất cho đến khi đạt số cụm mong muốn.

Cuối cùng, cách tiếp cận xác suất gom từng điểm dựa trên xác suất điểm đó thuộc về một cụm cụ thể.

Ta xem một số thuật toán phân cụm được dùng phổ biến nhất:
- K-means
- Fuzzy K-means
- Phân cụm phân cấp
- Hỗn hợp Gauss

K-means là ví dụ của thuật toán phân cụm loại trừ. Fuzzy K-means là ví dụ của thuật toán phân cụm chồng lấn. Phân cụm phân cấp là ví dụ của chính thuật toán phân cụm phân cấp. Hỗn hợp Gauss là ví dụ của thuật toán phân cụm xác suất.

Ta tìm hiểu rõ hơn từng phương pháp phân cụm nêu trên.

## **3. Phân cụm K-Means**

Phân cụm K-means hoạt động bằng cách phân loại một tập dữ liệu vào một số cụm xác định, số này được cố định từ trước.

Các bước của thuật toán k-means:
1. Ta bắt đầu bằng cách chọn k tâm cụm (centroid), trong đó k là số cụm.
2. Sau đó ta đặt k tâm cụm vào dữ liệu huấn luyện ở những vị trí ngẫu nhiên khác nhau.
3. Tiếp theo, ta tính khoảng cách từ mỗi điểm trong dữ liệu huấn luyện đến từng tâm cụm.
4. Các điểm được gom nhóm theo tâm cụm gần nhất.
5. Các điểm đã gom được tách riêng cùng với tâm cụm tương ứng. Tính trung bình của mỗi nhóm, rồi dời tâm cụm về vị trí trung bình vừa tính.
6. Ta lặp lại các bước cho đến khi đạt số nhóm k tối ưu.

Về mặt toán học, mục tiêu của thuật toán K-means là cực tiểu hóa một hàm mục tiêu, trong trường hợp này là hàm sai số bình phương:
$$J=\sum_{j=1}^k \sum_{i=1}^n ||x_i^{j} - C_j||^2$$
trong đó

$$||x_i^{j} - C_j||^2$$
là khoảng cách Euclid giữa các điểm dữ liệu $x_i$ và tâm cụm $C_j$.

Sau đây là các bước toán học của thuật toán:
1. Gọi $X = \{x_1, x_2, \cdots, x_n\}$ là các điểm dữ liệu và $V = \{v_1, v_2, \cdots, v_c\}$ là các tâm cụm.
2. Chọn ngẫu nhiên các tâm cụm.
3. Tính khoảng cách giữa các tâm cụm và từng điểm dữ liệu.
4. Gán các điểm dữ liệu cho tâm cụm gần chúng nhất.
5. Tính lại các tâm cụm mới bằng cách lấy trung bình của từng nhóm,
$$v_i = \left(\frac{1}{c_i}\right)\sum_{j=1}^{c_i} x_j$$
trong đó $c_i$ là số điểm dữ liệu trong cụm thứ i.
6. Tính khoảng cách giữa tâm cụm mới và từng điểm dữ liệu.
7. Nếu thành phần của các cụm không thay đổi thì dừng, nếu không thì lặp lại từ bước 3.

Vấn đề của thuật toán K-means là số cụm cần được chọn trước khi bắt đầu mô hình hóa. Ta có thể ước đoán số cụm bằng hệ số Silhouette hoặc phương pháp khuỷu tay.

### **3.1 Phân tích Silhouette**

Phân tích Silhouette được dùng trong phân cụm để khảo sát khoảng cách phân tách giữa các cụm. Ta dùng biểu đồ Silhouette để trực quan hóa khoảng cách của các điểm trong một cụm so với các điểm ở cụm lân cận. Thước đo này có khoảng giá trị là $[-1, 1]$.

Hệ số Silhouette gần $+1$ cho thấy các điểm dữ liệu ở xa các cụm lân cận. Hệ số bằng $0$ cho thấy các điểm dữ liệu rất gần các cụm lân cận. Hệ số âm cho thấy các điểm dữ liệu có thể đã bị gán sai cụm.

Hệ số Silhouette được tính như sau
$$S = \frac{b-a}{\max(a,b)}$$
Dưới đây là ví dụ về biểu đồ Silhouette.

### **3.2 Phương pháp khuỷu tay**

Ta có thể dựng biểu đồ khuỷu tay để tìm giá trị $k$ phù hợp dùng cho phân cụm.

Ta tìm số cụm (nhóm) tối ưu bằng cách vẽ đường biểu diễn Tổng bình phương trong cụm (WSS) theo số cụm. WSS được định nghĩa là tổng bình phương chênh lệch giữa mỗi điểm trong một cụm và tâm cụm của nó. Biểu đồ khuỷu tay được minh họa trong hình dưới đây:

Lưu ý rằng với thuật toán K-means, không có phương pháp tổng quát nào cho ra số cụm tối ưu duy nhất.

Bây giờ ta áp dụng K-means qua một ví dụ đơn giản.

```python
# Import packages

import matplotlib.pyplot as plt
import numpy as np
from sklearn.cluster import KMeans
from sklearn.datasets import make_blobs

np.random.seed(0)

%matplotlib inline
```

Ta tạo một tập dữ liệu gồm 4 cụm bằng hàm 'make_blobs' của thư viện scikit-learn và vẽ biểu đồ để trực quan hóa.

```python
X, y = make_blobs(centers=4, n_samples=2500)

fig = plt.figure(figsize=(8, 6))
plt.scatter(X[:, 0], X[:, 1], c=y)
plt.title("Dataset with 4 clusters", fontsize=20)
plt.xlabel("First feature")
plt.ylabel("Second feature")
plt.suptitle(
    "Fig. 5: Dataset Visualization", fontweight="bold", horizontalalignment="right"
)
plt.show()
```

Trong biểu đồ trên, ta thấy một số cụm giao nhau, và điều này đặt ra một bài toán thú vị: xem thuật toán của ta có phân biệt được chúng hay không.

Ta vẽ đường cong khuỷu tay để xác định liệu ta có thu được 4 cụm như kỳ vọng mà ta đã tạo ra hay không.

```python
wcss = []
for i in range(1, 11):
    kmeans = KMeans(
        n_clusters=i, init="k-means++", max_iter=300, n_init=10, random_state=42
    )
    kmeans.fit(X)
    wcss.append(kmeans.inertia_)

plt.rcParams["figure.figsize"] = (7, 5)
plt.plot(range(1, 11), wcss)
plt.title("K-Means Clustering(The Elbow Method)", fontsize=20)
plt.xlabel("K")
plt.ylabel("WCSS")
plt.suptitle("Fig. 6: The Elbow Method", fontweight="bold", horizontalalignment="right")
plt.show()
```

Lưu ý rằng sau giá trị $k = 4$, WCSS giảm rất chậm. Do đó giá trị $k$ của ta bằng 4.

```python
kmeans = KMeans(n_clusters=4, init="k-means++", max_iter=300, n_init=10, random_state=0)
ymeans = kmeans.fit_predict(X)

plt.rcParams["figure.figsize"] = (10, 10)
plt.title("Cluster of Dataset", fontsize=30)

plt.scatter(X[ymeans == 0, 0], X[ymeans == 0, 1], s=100, c="pink", label="Cluster 0")
plt.scatter(X[ymeans == 1, 0], X[ymeans == 1, 1], s=100, c="orange", label="Cluster 1")
plt.scatter(
    X[ymeans == 2, 0], X[ymeans == 2, 1], s=100, c="lightgreen", label="Cluster 2"
)
plt.scatter(X[ymeans == 3, 0], X[ymeans == 3, 1], s=100, c="red", label="Cluster 3")
plt.scatter(
    kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1], s=50, c="black"
)

plt.xlabel("First Feature")
plt.ylabel("Second Feature")
plt.legend()
# `plt.suptitle`('Fig 10: Clustering Results', fontsize=20, y = 1, ha = 'right')
plt.suptitle(
    "Fig. 7: Clustering Results", fontweight="bold", horizontalalignment="right"
)
plt.show()
```

Ta đã huấn luyện thuật toán K-means trên tập dữ liệu và thấy đầu ra cho thấy sự phân biệt rõ ràng giữa các cụm.

Trong phần tiếp theo, ta tìm hiểu cách tiếp cận kỳ vọng–cực đại hóa để tìm thành phần của các cụm.

### **3.3 Thuật toán k-Means: Kỳ vọng–Cực đại hóa**

Quy trình Kỳ vọng–Cực đại hóa (E-M) như sau.
1. Chọn ngẫu nhiên các tâm cụm.
2. Lặp lại các bước sau cho đến khi hội tụ:
- Bước E: các điểm được gán cho tâm cụm gần nhất.
- Bước M: tính trung bình của mỗi cụm và thay tâm cụm bằng giá trị trung bình đó.

Ta có thể quan sát thuật toán E-M hoạt động như bên dưới.

```python
from sklearn.datasets import make_blobs
from sklearn.metrics import pairwise_distances_argmin

X, y_true = make_blobs(n_samples=300, centers=4, cluster_std=0.60, random_state=0)

rng = np.random.RandomState(42)
centers = [0, 4] + rng.randn(4, 2)


def draw_points(ax, c, factor=1):
    ax.scatter(X[:, 0], X[:, 1], c=c, cmap="viridis", s=50 * factor, alpha=0.3)


def draw_centers(ax, centers, factor=1, alpha=1.0):
    ax.scatter(
        centers[:, 0],
        centers[:, 1],
        c=np.arange(4),
        cmap="viridis",
        s=200 * factor,
        alpha=alpha,
    )
    ax.scatter(centers[:, 0], centers[:, 1], c="black", s=50 * factor, alpha=alpha)


def make_ax(fig, gs):
    ax = fig.add_subplot(gs)
    ax.xaxis.set_major_formatter(plt.NullFormatter())
    ax.yaxis.set_major_formatter(plt.NullFormatter())
    return ax


fig = plt.figure(figsize=(15, 4))
gs = plt.GridSpec(
    4, 15, left=0.02, right=0.98, bottom=0.05, top=0.95, wspace=0.2, hspace=0.2
)
ax0 = make_ax(fig, gs[:4, :4])
ax0.text(
    0.98,
    0.98,
    "Random Initialization",
    transform=ax0.transAxes,
    ha="right",
    va="top",
    size=16,
)
draw_points(ax0, "gray", factor=2)
draw_centers(ax0, centers, factor=2)

for i in range(3):
    ax1 = make_ax(fig, gs[:2, 4 + 2 * i : 6 + 2 * i])
    ax2 = make_ax(fig, gs[2:, 5 + 2 * i : 7 + 2 * i])

    # E-step
    y_pred = pairwise_distances_argmin(X, centers)
    draw_points(ax1, y_pred)
    draw_centers(ax1, centers)

    # M-step
    new_centers = np.array([X[y_pred == i].mean(0) for i in range(4)])
    draw_points(ax2, y_pred)
    draw_centers(ax2, centers, alpha=0.3)
    draw_centers(ax2, new_centers)
    for i in range(4):
        ax2.annotate(
            "",
            new_centers[i],
            centers[i],
            arrowprops=dict(arrowstyle="->", linewidth=1),
        )

    # Finish iteration
    centers = new_centers
    ax1.text(
        0.95, 0.95, "E-Step", transform=ax1.transAxes, ha="right", va="top", size=14
    )
    ax2.text(
        0.95, 0.95, "M-Step", transform=ax2.transAxes, ha="right", va="top", size=14
    )


# Final E-step
y_pred = pairwise_distances_argmin(X, centers)
axf = make_ax(fig, gs[:4, -4:])
draw_points(axf, y_pred, factor=2)
draw_centers(axf, centers, factor=2)
axf.text(
    0.98,
    0.98,
    "Final Clustering",
    transform=axf.transAxes,
    ha="right",
    va="top",
    size=16,
)
plt.suptitle(
    "Fig. 8: Expectation-Maximization", fontweight="bold", horizontalalignment="right"
)
plt.show()
```

Các hạn chế của thuật toán E-M:
- Có thể không đạt được nghiệm tối ưu toàn cục.
- Phải chọn số cụm trước khi mô hình hóa.
- K-means sẽ chậm với dữ liệu lớn.
- K-means chỉ hoạt động với ranh giới cụm tuyến tính.

## **4. Phân cụm Fuzzy K-Means**

Phân cụm mờ giống thuật toán K-means, điểm khác là mỗi điểm có một mức độ (xác suất) thuộc về từng cụm trong tập dữ liệu. Trong trường hợp này, một điểm có thể thuộc 2 cụm trở lên nếu nó nằm ở giữa các cụm.

Các quá trình khác trong K-means như khởi tạo ngẫu nhiên, lặp và điều kiện dừng thuật toán đều giống nhau ở fuzzy K-means. Do đó, các cụm được xử lý như các phân phối xác suất. Nếu chuyển hàm xác suất thành phân phối nhị phân, cho giá trị 1 với các điểm dữ liệu gần tâm cụm và 0 với các điểm còn lại, fuzzy K-means trở thành K-means thông thường.

Quy trình của thuật toán fuzzy K-means như sau;

- Trước hết, cố định số cụm.
- Khởi tạo ngẫu nhiên các tâm cụm và tính xác suất các điểm dữ liệu thuộc về các cụm khác nhau.
- Khi đã biết xác suất mỗi điểm dữ liệu thuộc về một cụm, ta tính lại các tâm cụm mới bằng cách lấy trung bình của các cụm.
$$v_k (n+1) = \frac{\sum_{x_i\in k} x_i P(v_k|x_i)^b}{\sum_{x_i\in k}P(v_k|x_i)^b}$$
- Thuật toán dừng khi hội tụ hoặc khi đạt số vòng lặp quy định.

## **5. Phân cụm phân cấp**

Trong phân cụm phân cấp, ta định nghĩa một ma trận khoảng cách bằng liên kết (linkage) với các tham số sau
- Phương pháp (Method): để tính khoảng cách giữa các cụm.
- Thước đo (Metric): thước đo khoảng cách được dùng.
- Sắp xếp tối ưu (Optimal ordering): để sắp thứ tự các điểm dữ liệu.

Các loại phương pháp dùng trong phân cụm phân cấp:

1. Single: tìm khoảng cách giữa các điểm gần nhất ở các cụm khác nhau.
2. Complete: tìm khoảng cách giữa các điểm xa nhất ở các cụm khác nhau.
3. Average: tìm khoảng cách giữa trung bình cộng của các cụm.
4. Centroids: tìm khoảng cách giữa trung bình hình học của các cụm.
5. Median: tìm khoảng cách giữa trung vị của các cụm.
6. Ward: tìm khoảng cách dựa trên tổng bình phương.

Dưới đây là các đoạn mã Python minh họa cách triển khai những phương pháp khác nhau.

### **5.1 Phương pháp Ward**

```python
import pandas as pd
from sklearn.datasets import make_blobs

X1, y1 = make_blobs(
    n_samples=50, centers=[[4, 4], [-2, -1], [1, 1], [10, 4]], cluster_std=0.9
)
df = pd.DataFrame(X1, columns=["X", "y"])
df.head()
```

```python
import seaborn as sns
from scipy.cluster.hierarchy import fcluster, linkage

# Use the linkage()
distance_matrix = linkage(df[["X", "y"]], method="ward", metric="euclidean")

# Assign cluster labels
df["cluster_labels"] = fcluster(distance_matrix, 2, criterion="maxclust")

# Plot clusters
sns.scatterplot(x="X", y="y", hue="cluster_labels", data=df)
plt.suptitle("Fig. 9: Ward Method", fontweight="bold", horizontalalignment="right")
plt.show();
```

### **5.2 Phương pháp Single**

```python
# Use the linkage()
distance_matrix = linkage(df[["X", "y"]], method="single", metric="euclidean")

# Assign cluster labels
df["cluster_labels_2"] = fcluster(distance_matrix, 2, criterion="maxclust")

# Plot clusters
sns.scatterplot(x="X", y="y", hue="cluster_labels_2", data=df)
plt.suptitle("Fig. 10: Single Method", fontweight="bold", horizontalalignment="right")
plt.show();
```

### **5.3 Phương pháp Complete**

```python
# Use the linkage()
distance_matrix = linkage(df[["X", "y"]], method="complete", metric="euclidean")

# Assign cluster labels
df["cluster_labels_3"] = fcluster(distance_matrix, 2, criterion="maxclust")

# Plot clusters
sns.scatterplot(x="X", y="y", hue="cluster_labels", data=df)
plt.suptitle("Fig. 11: Complete Method", fontweight="bold", horizontalalignment="right")
plt.show();
```

### **5.4 Biểu đồ cây (Dendrogram)**

Dendrogram giúp ta trực quan hóa cách các cụm được hình thành, và từ đó ta có thể chọn số cụm trong tập dữ liệu. Dưới đây là đoạn mã đơn giản cho ra biểu đồ dendrogram.

```python
from scipy.cluster.hierarchy import dendrogram

# Create a dendrogram
dn = dendrogram(distance_matrix)
plt.suptitle("Fig. 11: Dendrogram", fontweight="bold", horizontalalignment="right")
plt.show()
```

Tóm lại, các bước của phân cụm phân cấp như sau:
1. Ta coi mỗi điểm trong tập dữ liệu là một cụm riêng.
2. Sau đó ta tính khoảng cách Euclid giữa các tâm của mọi cụm trong tập dữ liệu.
3. Ta nối các cụm gần nhất lại với nhau.
4. Lặp lại bước 2 và 3 cho đến khi chỉ còn một cụm chứa mọi điểm dữ liệu.
5. Vẽ dendrogram thể hiện quá trình tiến hóa của cấu trúc phân cấp, cho thấy các cụm được sắp xếp từ trên xuống dưới.
6. Đến lúc này, việc quyết định tạo cụm ở mức nào là tùy ở ta.

## **6. Mô hình hỗn hợp Gauss (GMM)**

Đây là cách tiếp cận dựa trên mô hình, dùng một số mô hình để tạo ra các cụm và cố gắng tối ưu độ khớp của mô hình với dữ liệu.

Ưu điểm của loại phân cụm này gồm khả năng một điểm thuộc về nhiều cụm, ước lượng mật độ cho từng cụm, sự sẵn có của các phương pháp suy luận thống kê vững chắc để thực hiện phân cụm, và sự tự do chọn phân phối thành phần của mô hình.

$K$ phân phối thành phần tạo nên mô hình hỗn hợp được kết hợp lại để cho ra $$f(x) = \sum_{k= 1}^K a_k f_k(x)$$
trong đó $a_k$ là đóng góp của thành phần thứ $k$ trong việc tạo nên $f(x)$.

Giống thuật toán K-means ở trên, GMM cũng được dùng để tìm số cụm trong một tập dữ liệu. Ta xét ví dụ dưới đây.

```python
%matplotlib inline
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

# Generate some data
from sklearn.datasets import make_blobs

sns.set()

X, y_true = make_blobs(n_samples=400, centers=4, cluster_std=0.60, random_state=0)
X = X[:, ::-1]  # flip axes for better plotting
```

Để trực quan hóa số cụm, ta khớp mô hình hỗn hợp Gauss với số cụm được chỉ định trước.

```python
from sklearn import mixture

gmm = mixture.GaussianMixture(n_components=4).fit(X)
labels = gmm.predict(X)
plt.scatter(X[:, 0], X[:, 1], c=labels, s=40, cmap="viridis")
plt.suptitle(
    "Fig. 12: Clusters from GMM", fontweight="bold", horizontalalignment="right"
)
plt.show();
```

Mô hình hỗn hợp Gauss chứa một mô hình xác suất cho biết xác suất một điểm thuộc về một cụm.

```python
probs = gmm.predict_proba(X)
print(probs[:5].round(3))
```

Mô hình hỗn hợp Gauss, giống thuật toán k-means, dùng cách tiếp cận kỳ vọng–cực đại hóa như sau.
1. Ta chọn một điểm xuất phát ngẫu nhiên.
2. Lặp lại các bước sau cho đến khi hội tụ.
- Bước E: tìm trọng số của mỗi điểm thuộc về một cụm nhất định.
- Bước M: cập nhật vị trí và hình dạng của từng cụm dựa trên trọng số của các điểm dữ liệu.

GMM đôi khi bỏ lỡ nghiệm tối ưu toàn cục, vì vậy nên dùng các cách khởi tạo ngẫu nhiên khác nhau.

### **6.1 Số cụm**

GMM là một mô hình sinh (generative model) nên cho ta một kỹ thuật tự nhiên để tìm số cụm tối ưu. Vì mô hình sinh cung cấp phân phối xác suất cho một tập dữ liệu, ta có thể dùng các tiêu chí phân tích như *Tiêu chuẩn thông tin Akaike (AIC)* hoặc *Tiêu chuẩn thông tin Bayes (BIC)* để giảm khả năng mô hình bị quá khớp. Ta cũng có thể dùng kiểm định chéo để đạt cùng mục đích.

Trong ví dụ dưới đây, ta dùng AIC và BIC.

```python
from sklearn.datasets import make_moons

Xmoon, ymoon = make_moons(200, noise=0.05, random_state=0)
plt.scatter(Xmoon[:, 0], Xmoon[:, 1])

n_components = np.arange(1, 21)
models = [
    mixture.GaussianMixture(n, covariance_type="full", random_state=0).fit(Xmoon)
    for n in n_components
]

plt.plot(n_components, [m.bic(Xmoon) for m in models], label="BIC")
plt.plot(n_components, [m.aic(Xmoon) for m in models], label="AIC")
plt.legend(loc="best")
plt.xlabel("n_components")
plt.suptitle(
    "Fig. 13: Optimal Clusters", fontweight="bold", horizontalalignment="right"
);
```

## **7. DBSCAN**

DBSCAN là viết tắt của Density-Based Spatial Clustering of Applications with Noise (phân cụm không gian dựa trên mật độ có xử lý nhiễu). Phương pháp này hoạt động bằng cách nhận diện các vùng có mật độ điểm dày so với các vùng thưa. Logic ở đây là các điểm thuộc một cụm phải gần các điểm cùng cụm.

Phương pháp được xác định bằng hai tham số:
1. **Epsilon**: xác định bán kính trong đó có đủ số điểm.
2. **Số mẫu tối thiểu**: xác định số điểm tối thiểu ta muốn có trong một cụm.

Dưới đây là cách triển khai thuật toán DBSCAN.

```python
import matplotlib.pyplot as plt
import numpy as np
from sklearn.cluster import DBSCAN
from sklearn.datasets import make_blobs
from sklearn.preprocessing import StandardScaler

%matplotlib inline
```

```python
def createDataPoints(centroidLocation, numSamples, clusterDeviation):
    # Create random data and store in feature matrix X and response vector y.
    X, y = make_blobs(
        n_samples=numSamples, centers=centroidLocation, cluster_std=clusterDeviation
    )

    # Standardize features by removing the mean and scaling to unit variance
    X = StandardScaler().fit_transform(X)
    return X, y
```

```python
X, y = createDataPoints([[4, 3], [2, -1], [-1, 4]], 1500, 0.5)
```

```python
epsilon = 0.3
minimumSamples = 7
db = DBSCAN(eps=epsilon, min_samples=minimumSamples).fit(X)
labels = db.labels_
labels
```

```python
# First, create an array of booleans using the labels from db.
core_samples_mask = np.zeros_like(db.labels_, dtype=bool)
core_samples_mask[db.core_sample_indices_] = True
core_samples_mask
```

```python
# Number of clusters in labels, ignoring noise if present.
n_clusters_ = len(set(labels)) - (1 if -1 in labels else 0)
n_clusters_
```

```python
# Remove repetition in labels by turning it into a set.
unique_labels = set(labels)
unique_labels
```

```python
# Create colors for the clusters.
colors = plt.cm.Spectral(np.linspace(0, 1, len(unique_labels)))
colors
```

```python
# Plot the points with colors
for k, col in zip(unique_labels, colors):
    if k == -1:
        # Black used for noise.
        col = "k"

    class_member_mask = labels == k

    # Plot the datapoints that are clustered
    xy = X[class_member_mask & core_samples_mask]
    plt.scatter(xy[:, 0], xy[:, 1], s=50, c=col, marker="o", alpha=0.5)

    # Plot the outliers
    xy = X[class_member_mask & ~core_samples_mask]
    plt.scatter(xy[:, 0], xy[:, 1], s=50, c=col, marker="o", alpha=0.5)
plt.show()
```
